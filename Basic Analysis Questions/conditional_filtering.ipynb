# Conditional Filtering

```python
import pandas as pd
```

```python
df = pd.read_csv('Dataset/Car_sales.csv')
df.head()
```

```python
# Find car details whose model year is 2017.
filter_car = df['year'] == 2017
df[filter_car]
# or
'''
df[df['year]==2017]
'''
```

```python
# now to select only specific col let here be name year selling price where model year is 2017.
df[df['year']==2017][['name','year','selling_price']]
# or
# df[filter_car][['name','year','selling_price']]
```

```python
# Find car details whose price is in range of 300000 to 600000.
# for continuous data like price, salary, temperature
# in such data which has range we have to use either and ,or between(not or).
# here in pandas and is denoted by &.
# using and(&)
car_price = df[(df['selling_price'] >= 300000) & (df['selling_price']<=600000)]
car_price
```

```python
# using between
car_price_between=df[df['selling_price'].between(300000,600000)]
car_price_between
```

```python
# in large dataset & work faster as & works fully in boolean but between first compares values and converts to boolean
# in query optimization must useand(&).
# in between both values are inclusive.
```

```python
# for categorical data we must use or(not and), or in(.isin())
```

```python
# to find unique values
df['fuel'].unique() #to get names of unique values
```

```python
# to find number of unique values.
df['fuel'].nunique()
```

```python
# Find cars whose emission fuel type is Petrol or CNG.
car_fuel=df[(df['fuel']=='Petrol')|(df['fuel']=='CNG')]
car_fuel
# here \is CNG
```

```python
car_fuel_in = df[df['fuel'].isin(['Petrol','CNG'])]
car_fuel_in
```

```python
unique_fuel=car_fuel['fuel'].unique()
unique_fuel
```

```python
# Find car details whose transmission mode is Automatic, fuel type is Petrol/Electric and which is driven for around 20000 to 35000.
car_details = df[(df['transmission']=='Automatic')
                 & ((df['fuel']=='Petrol') | (df['fuel']=='Electric')) & 
                 ((df['km_driven']>=20000) & (df['km_driven']<=35000))]
car_details
```

```python
# to get only specific col
car_details[['name','selling_price']]
```
