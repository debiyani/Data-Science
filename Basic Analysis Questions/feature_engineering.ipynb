# Convert Continuous data into Categorical Column

```python
'''
0-5-> Low Absence
6-10-> Medium Absence
10>-> High Absence
'''
```

```python
df['Absences'].unique()
```

```python
def absenceCategory(absence):
    if absence>0 and absence<=5:
        return 'Low Absence'
    elif absence>5 and absence<=10:
        return 'Medium Absence'
    elif absence>10:
        return 'High Absence'
    else:
        return 'Error: Negative Absence'
```

```python
df['Absence_Label'] = df['Absences'].apply(absenceCategory)
```

```python
df
```

# Data Encoding

## Manual Encoding

```python
data = {
    'Low Absence': 0,
    'Medium Absence': 1,
    'High Absence': 2
}
```

```python
df['Absence_Id'] = df['Absence_Label'].map(data)
```

```python
df
```

```python
def encodeData(dataframe, col_name, *args):
    encode_data = {}
    for idx, val in enumerate(args):
        encode_data[val] = idx
    dataframe[f'{col_name}_id'] = dataframe[col_name].map(encode_data)
    print(f'{col_name} data encoded.')
```

```python
encodeData(df, 'Absence_Label','Low Absence', 'Medium Absence', 'High Absence')
```
