```python
import matplotlib.pyplot as plt
```

```python
company = {
    'employee_name':['Bob','Ram','Shyam','Jacob','John'],
    'salary':[25000,18000,30000,49000,45000],
    'bonus':[5000,1200,15000,8000,18000]
}
```

```python
plt.plot(company['employee_name'], company['salary'])
plt.xlabel('Employee Name')
plt.ylabel('Salary')
plt.title('Salary distribution of Employees.')
```

```python
plt.plot(company['employee_name'], company['salary'])
plt.xlabel('Employee Name')
plt.ylabel('Salary')
plt.title('Salary distribution of Employees.')
plt.grid(linestyle='dashdot', alpha=0.4) #alpha is opacity
plt.show()
```

```python
plt.plot(company['employee_name'], company['salary'],color='#000fff', marker='o', linestyle = 'dotted')
plt.xlabel('Employee Name')
plt.ylabel('Salary')
plt.title('Salary distribution of Employees.')
plt.grid(linestyle='dashdot', alpha=0.4) #alpha is opacity
plt.show()
```

```python
def formatData(val):
    if val>1e9:
        return f'{val/1e9:,.2f}B'
    elif val>1e6:
        return f'{val/1e6:,.2f}M'
    if val>1e3:
        return f'{val/1e3:,.2f}K'
    else:
        return f'{val}'
    
def plotData(dataframe, col_name, prefix='', ha_point='center', va_point='bottom'):
    for idx, val in enumerate(dataframe[col_name]):
        formatted_val = f'{prefix}'+formatData(val)
        plt.text(idx, val, formatted_val, ha=ha_point , va=va_point )
```

```python
plt.plot(company['employee_name'], company['salary'],color='#000fff', marker='o', linestyle = 'dotted', label='Salary')
plt.plot(company['employee_name'],company['bonus'],color='red', marker='*',label='BOnus')
plt.xlabel('Employee Name')
plt.ylabel('Salary')
plt.title('Salary distribution of Employees.')
plt.grid(linestyle='dashdot', alpha=0.4) #alpha is opacity

# show salary value in plot
# ha= horizontal alignment = left, right, center
# left garda right ma aauxa and right garda left ma also top garda bottom and so on center garda bichma
# va = vertical alignment = top, bottom, center
# for idx, val in enumerate(company['salary']):
#     formatted_val = formatData(val)
#     plt.text(idx, val, formatted_val, ha='center' , va='bottom' )
plotData(company, 'salary', 'Rs.')
plt.show()
```

```python
df= pd.read_csv('cleaned_hr.csv')
df.head()
```

# Pie Chart

```python
perf = df.groupby('PerformanceScore')['EmpID'].count().reset_index()
perf
```

```python
plt.pie(
    perf['EmpID'],
    labels = perf['PerformanceScore'],
    autopct = '%1.1f%%',
    explode=(0.3,0,0,0)
    # default explode 0,0,0,0
    # explode le bahira nikalxa value lai
)
plt.title('Total Employees per performance score')
plt.show()
```

## Donut chart 
A type of pie chart

```python
plt.pie(
    perf['EmpID'],
    labels = perf['PerformanceScore'],
    autopct='%1.1f%%',
    # width edge properties
    # props ma dict value janxa
    wedgeprops ={'width': 0.5},
    # width 1 pie chart
    pctdistance=0.8,
    explode=(0.2,0,0,0)
)
plt.show()
```

```python
plt.pie(
    perf['EmpID'],
    labels = perf['PerformanceScore'],
    autopct='%1.1f%%',
    # width edge properties
    # props ma dict value janxa
    wedgeprops ={'width': 0.5},
    # width 1 pie chart
    pctdistance=0.8,
    explode=(0.2,0,0,0),
    textprops={'color':'red'}
)
plt.title('Total Employees per performance score',fontweight='bold',fontfamily='cursive', fontstyle='italic', color='teal', fontsize='large')
plt.show()
```

# Histogram

```python
filter_salary = df[df['Salary']<80000]
```

```python
filter_salary['Salary'].skew()
```

```python
plt.figure(figsize=(6,10))
plt.hist(filter_salary['Salary'], edgecolor = 'black')
plt.show()
```

```python
# KDE(Kernal Density Estimation) lines
# mean median lines
# axv line(axis vertial line)
# all these sns seaborn ma milxa but not in matplotlib sns ma bins pani hunxa
```

```python
mean_salary=filter_salary['Salary'].mean()
median_salary = filter_salary['Salary'].median()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(filter_salary['Salary'], kde=True, bins=10)
plt.axvline(mean_salary, color='red', linestyle='dashed', label='Mean salary')
plt.axvline(median_salary, color='yellow', label='Median salary')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(filter_salary['Salary'], kde=True, bins=100)
plt.show()
```
