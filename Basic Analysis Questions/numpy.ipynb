## Reshape and ravel

```python
'''
Reshape-> Change any dimension of array to any other dimesnsion
Ravel-> Change any dimension array to 1d array/ flatten in ml or scikit learn

'''
```

```python
reshape_numbers=numbers.reshape(10,6)
reshape_numbers
```

```python
ravel_numbers=np.ravel(reshape_numbers)
ravel_numbers
```
