# Data Cleaning

```python
'''
1. Handle missing/null values
    a. Fill Null Values
    
    -> .fillna()
     
        -Continuous Data-> Mean, Median
        -Discrete/ Categorical Data-> Mode
    -------------------------------------------------------------------------------------------------------------------------    
    Continuous data
    -------------------------------------------------------------------------------------------------------------------------   
    -> to fill, we first need to find skewness using .skew()
    
    ->skewness 0-0.5-> fill mean value
    -> skewness <0(Negative skewness) or > skewness 0.5(Positive Skewness)-> fill median value
    
    -> or using mean and median value of a column
    
        -> if both mean equal/near equal, fill mean value
        -> if mean median not equal and vast difference, fill median value
   -----------------------------------------------------------------------------------------------------------------------------
    b. Drop Null Values
    ->.dropna()
---------------------------------------------------------------------------------------------------------------------------------   
2. Remove unwanted data
    -> Remove whitespace
        -> .strip(), .lstrip(), .rstrip()
        
    -> Remove Regular Expressions(regex) if needed/ Remove unwanted texts
        -> .replace()
---------------------------------------------------------------------------------------------------------------------------------
3. Change Data types
-> .astype()
--------------------------------------------------------------------------------------------------------------------------------
4. Handling duplicate Records
---------------------------------------------------------------------------------------------------------------------------------
-> Find duplicate records
    -> .duplicated()-> checks entire row wise
    -> .duplicated(subset)-> checks col wise

-> To remove duplicate records
    -> .drop_duplicates()

'''
```

## Removing whitespace for multiple col

```python
def removeWhitespace(dataframe, columns):
    for col in columns:
        dataframe[col] = dataframe[col].str.strip()
    print('White spaces removed.')
```

```python
removeWhitespace(hr, columns)
```

## Replace unwanted text, regex from data

```python
'''
str.replace(find, replace_value)

-> for text /char/digits
    -str.replace(find, replace_value)
-> for special characters
    -str.replace(find, replace_value, regex=False/True(default true))
    
'''
```

```python
hr['Employee_Name'] = hr['Employee_Name'].str.replace(',', '', regex= False)
```
