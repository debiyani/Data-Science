# Loc and Iloc

```python
'''
->used for slicing both row and col
LOC-> Location-> extract data using alphabetical index
ILOC-> Integer Location-> Extract data using numeric index
# No step in loc and iloc
'''
```

```python
products.iloc[1:3,1:]
```

```python
products.loc['B':'C','Brand':]
```

```python
products.iloc[
    [0,1,3],
    [0]
]
```

```python
products.loc[
    ['A','B','D'],
    ['Product Name']
]
```
